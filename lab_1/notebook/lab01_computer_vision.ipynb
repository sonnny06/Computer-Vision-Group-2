```python
import cv2
import matplotlib.pyplot as plt
```

```python
from pathlib import Path

# 1. Đọc ảnh từ máy tính; thay đường dẫn này để dùng ảnh khác.
image_path = Path("../data/input/01_portrait_grace_hopper.jpg")
if not image_path.exists():
    image_path = Path("lab_1/data/input/01_portrait_grace_hopper.jpg")
if not image_path.is_file():
    raise FileNotFoundError(f"Không tìm thấy ảnh: {image_path}. Hãy sửa image_path.")

image_bgr = cv2.imread(str(image_path))
if image_bgr is None:
    raise ValueError(f"OpenCV không đọc được ảnh: {image_path}")
height, width = image_bgr.shape[:2]
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
print(f"Đã đọc ảnh {width} x {height}: {image_path}")

# 2. Chuyển đổi màu: grayscale, HSV và LAB.
gray_image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)
hsv_image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)
lab_image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2LAB)
hsv_preview = cv2.cvtColor(hsv_image, cv2.COLOR_HSV2RGB)
lab_preview = cv2.cvtColor(lab_image, cv2.COLOR_LAB2RGB)

# 3. Cắt vùng trung tâm, resize theo tỷ lệ và theo kích thước cố định.
x1, x2 = width // 4, max(width // 4 + 1, 3 * width // 4)
y1, y2 = height // 4, max(height // 4 + 1, 3 * height // 4)
cropped_image = image_bgr[y1:y2, x1:x2]
resized_by_ratio = cv2.resize(image_bgr, None, fx=0.5, fy=0.5)
resized_fixed = cv2.resize(image_bgr, (640, 480))

# 4. Vẽ đường thẳng, hình chữ nhật, hình tròn và chữ lên bản sao ảnh.
annotated_image = image_bgr.copy()
cv2.line(annotated_image, (0, 0), (width - 1, height - 1), (0, 255, 0), 3)
cv2.rectangle(annotated_image, (width // 10, height // 10),
              (max(width // 10 + 1, 4 * width // 10), max(height // 10 + 1, 4 * height // 10)),
              (255, 0, 0), 3)
cv2.circle(annotated_image, (width // 2, height // 2), max(1, min(width, height) // 8), (0, 0, 255), 3)
cv2.putText(annotated_image, "Computer Vision", (10, min(height - 5, 35)),
            cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 0), 2, cv2.LINE_AA)

# Lưu kết quả dưới định dạng PNG.
output_path = image_path.with_name(f"{image_path.stem}_processed.png")
if not cv2.imwrite(str(output_path), annotated_image):
    raise OSError(f"Không thể lưu ảnh tại: {output_path}")
print(f"Đã lưu ảnh đã xử lý: {output_path}")

# Hiển thị ảnh gốc và các kết quả xử lý.
previews = [
    (image_rgb, "Ảnh gốc (RGB)"), (gray_image, "Grayscale"),
    (hsv_preview, "HSV (hiển thị RGB)"), (lab_preview, "LAB (hiển thị RGB)"),
    (cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB), "Vùng đã cắt"),
    (cv2.cvtColor(resized_by_ratio, cv2.COLOR_BGR2RGB), "Resize tỷ lệ 0.5"),
    (cv2.cvtColor(resized_fixed, cv2.COLOR_BGR2RGB), "Resize 640 x 480"),
    (cv2.cvtColor(annotated_image, cv2.COLOR_BGR2RGB), "Đã vẽ và lưu PNG"),
]
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for axis, (preview, title) in zip(axes.flat, previews):
    axis.imshow(preview, cmap="gray" if preview.ndim == 2 else None)
    axis.set_title(title)
    axis.axis("off")
plt.tight_layout()
plt.show()
```
